```python
import pandas as pd 

doc = pd.read_csv( "/Users/mac/Python_basic/Python_won/pandas/00_PANDASPLOTLY_FUNCODING_FIRSTDATA/COVID-19-master/csse_covid_19_daily_reports/04-01-2020.csv", encoding="utf-8-sig")
```

```python
doc2 = doc.corr(numeric_only=True)
```

```python
doc2
```

```python
import plotly.express as px 

fig_px = px.imshow(
    doc2,
    text_auto=".2f",
    aspect = "auto",
    title="Heatmap"
)

fig_px.show()
```

```python
import plotly.graph_objects as go

fig = go.Figure()

fig.add_trace(
    go.Heatmap(
        x=doc2.columns,
        y=doc2.index,
        z=doc2.values,
        colorscale="Cividis",
        colorbar={
            "title":"Value"
        },
        text=doc2.values,
        texttemplate="%{z:.2f}",
        hovertemplate="x: %{x}<br>Y:%{y}<br>Value : %{z:.2f}<br>Custom Info"
    )
)

fig.update_layout(
    title="Heatmap"
)

fig.show()
```
