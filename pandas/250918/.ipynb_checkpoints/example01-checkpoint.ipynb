```python
#함수 선언문 
def add(a,b) :
    return a + b 

# 함수 호출 
print(add(1,2))
    
```

```python
# lambda 식 
# 함수의 이름을 없애는 대신, 해당 함수를 여튼 어떠한 변수 안애는 넣어줘야함 

add = lambda a, b : a + b 

print(add(1,2))
```
