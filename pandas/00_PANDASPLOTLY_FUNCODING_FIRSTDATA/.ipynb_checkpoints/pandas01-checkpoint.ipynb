```python
#데이터프레임에서 중복 행 확인/제거

import pandas as pd 

PATH = "COVID-19-master/csse_covid_19_data/UID_ISO_FIPS_LookUp_Table.csv"
```

```python
doc = pd.read_csv(PATH, encoding="utf-8-sig")
doc = doc[["iso2", "Country_Region"]]
doc
```

```python
doc.duplicated()
```

```python
doc[doc.duplicated()]
#현재 여러번 중복해서 사용된 값등을 1개만 남기고 나머지는 제거 해야함
#그러면 어떤 값을 지우고 살려야할지 
```

```python
doc = doc.drop_duplicates(subset="Country_Region", keep="last")
doc
```
