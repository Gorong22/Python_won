```python
# 파이썬에서 자주사용하는 대표적인 머신러닝 라이브러리 모듈 -> 실습예제 다양, 머신러닝 모델

# import sklearn
# sklearn.datasets.load_digits()

from sklearn.datasets import load_digits # 손글씨 형태의 샘플링 이미지 데이터 세트
digits = load_digits()

print(digits)
```

```python
digits["data"] # 각 이미지를 숫자화한 데이터
```

```python
digits["target"] # 각 이미지에 대한 정답
```

```python
from matplotlib import pyplot as plt # 파이썬에서 그림(그래프, 이미지)을 그리게 해주는 라이브러리 모듈
# 해당 라이브러리를 통해서 그려진 그림을 주피터노트북 > 셀에 바로 표기될 수 있도록 작성하는 코드
%matplotlib inline

fig = plt.figure(figsize=(6, 6)) # 앞으로 그릴 그림의 전체사이즈 정의한 값 (6, 6인치)
fig.subplots_adjust(left=0, right=1, bottom=0, top=1, hspace=0.05, wspace=0.05)

for i in range(64) :
    ax = fig.add_subplot(8, 8, i + 1, xticks=[], yticks=[])
    ax.imshow(digits.images[i], cmap=plt.cm.binary, interpolation="nearest")
    ax.text(0, 7, str(digits.target[i]))

plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test  = train_test_split(digits.data, digits.target)
```

```python
from sklearn.naive_bayes import GaussianNB

OneOfNavieBayesianModels = GaussianNB()
```

```python
OneOfNavieBayesianModels.fit(X_train, y_train)
```

```python
predicted = OneOfNavieBayesianModels.predict(X_test)
```

```python
fig = plt.figure(figsize=(6,6))
fig.subplots_adjust(left=0, right=1, bottom=0, top=1, hspace=0.05, wspace=0.05)

for i in range(64):
    ax = fig.add_subplot(8, 8, i + 1, xticks=[], yticks=[])
    ax.imshow(X_test.reshape(-1, 8, 8)[i], cmap=plt.cm.binary, interpolation="nearest")

    if predicted[i] == y_test[i]:
        ax.text(0, 7, str(predicted[i]), color="green")
    else:
        ax.text(0, 7, str(predicted[i]), color="red")

plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn import datasets, svm
from sklearn.model_selection import train_test_split

# 1) 데이터 불러오기
digits = datasets.load_digits()

X = digits.images.reshape(len(digits.images), -1)
y = digits.target

# 2) train/test 분리
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=0, stratify=y
)

# 3) 분류 모델 학습
clf = svm.SVC(gamma=0.001)
clf.fit(X_train, y_train)

# 4) 예측
predicted = clf.predict(X_test)

# 5) 그림 설정
fig = plt.figure(figsize=(6, 6))
fig.subplots_adjust(left=0, right=1, bottom=0, top=1,
                    hspace=0.05, wspace=0.05)

# 6) 64개만 출력
for i in range(64):
    ax = fig.add_subplot(8, 8, i + 1, xticks=[], yticks=[])
    ax.imshow(X_test.reshape(-1, 8, 8)[i],
              cmap=plt.cm.binary, interpolation="nearest")

    # 정답이면 초록, 오답이면 빨강
    if predicted[i] == y_test[i]:
        ax.text(0, 7, str(predicted[i]), color="green")
    else:
        ax.text(0, 7, str(predicted[i]), color="red")

plt.show()
```
