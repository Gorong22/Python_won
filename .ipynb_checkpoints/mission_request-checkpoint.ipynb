```python
import requests
from bs4 import BeautifulSoup

url = "https://books.toscrape.com/catalogue/category/books/sports-and-games_17/index.html"
res = requests.get(url)
soup = BeautifulSoup(res.content, "html.parser")

category = soup.select_one("div.page-header.action h1").text.strip()
print("카테고리:", category)

# 책 정보 추출
data = soup.select("li.col-xs-6.col-sm-4.col-md-3.col-lg-3")

for i, book in enumerate(data, start=1):
    title = book.select_one("h3 a")["title"].strip() 
    price = book.select_one("p.price_color").text.strip()
    stock = book.select_one("p.instock.availability").text.strip()
    print(f"{i}. {title} | {price} | {stock}")
    
```

```python
#mysql 삽입 
import requests
from bs4 import BeautifulSoup
import pymysql


url = "https://books.toscrape.com/catalogue/category/books/sports-and-games_17/index.html"
res = requests.get(url)
soup = BeautifulSoup(res.content, "html.parser")


category = soup.select_one("div.page-header.action h1").text.strip()
print("카테고리:", category)


data = soup.select("li.col-xs-6.col-sm-4.col-md-3.col-lg-3")


conn = pymysql.connect(
    host='localhost',
    user='root',
    password='Wonsil929621!@#',
    db='bookstoreDB',
    charset='utf8mb4'
)
cursor = conn.cursor()


sql = """
    INSERT INTO booklist (category, title, price, stock)
    VALUES (%s, %s, %s, %s)
"""


for i, book in enumerate(data, start=1):
    title = book.select_one("h3 a")["title"].strip()
    price_raw = book.select_one("p.price_color").text.strip()
    price_value = float(price_raw.replace("£", ""))   
    stock = book.select_one("p.instock.availability").text.strip()

    print(f"{i}. {title} | {price_value} | {stock}")  
    cursor.execute(sql, (category, title, price_value, stock))


conn.commit()
conn.close()

print("booklist 테이블에 데이터 삽입 완료!")
```
