```python
#11번가에서 '캠핑용품' 검색 후 -> 인기순 카테고리 클릭, 10페이지 까지 상품명 크롤링

from selenium import webdriver

from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time 

service = Service(ChromeDriverManager().install())
options = Options()

options.add_argument('--window-size=1920x1080')
options.add_argument('--start-maximized')
options.add_argument('--user-agent = Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36')
options.add_argument('--lang=ko_KR')

driver = webdriver.Chrome(service=service, options=options)
url = "https://www.11st.co.kr/page/main/home"
driver.get(url)

popup_button = driver.find_element(By.CSS_SELECTOR, "button.btn_close")
time.sleep(2)
popup_button.click()



search_button = driver.find_element(By.CSS_SELECTOR, "div.c_gnb_search.c_gnb_search_tab input.search_text")
search_button.click()
time.sleep(2)

search_button.send_keys("캠핑용품")
search_button.send_keys(Keys.ENTER)
time.sleep(3)

filter_button = driver.find_element(By.CSS_SELECTOR, "button.btn_icon.select")
time.sleep(1)
filter_button.click()

famus_button = driver.find_element(By.XPATH, ".//ul[@class='dropdown_list']//button[contains(text(), '11번가 인기순')]")
time.sleep(1)
famus_button.click()



for page_num in range(1, 11):
    print(driver.current_url)

    page = driver.find_elements(By.XPATH, ".//ul[contains(@class, 'c-search-list__container')]/dl")

    for pages in page :
        print(pages.text)

    driver.execute_script("window.scrollTo(0, document.body.scrollHeight)")
    num_button = driver.find_elements(By.XPATH, "//li[@class='active']/button")
    num_button[-1].click()

    time.sleep(10)

```
