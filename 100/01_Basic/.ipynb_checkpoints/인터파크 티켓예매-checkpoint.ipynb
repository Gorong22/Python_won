```python
#문제 4.
#인터파크 티켓 예매 페이지에서 인기 공연 데이터를 수집하고 공연 카테고리별 소비 분석
#https://nol.interpark.com/ticket
#1) 공연명, 날짜, 장소, 가격
#2) MySQL 저장
#3) SQL로 분석해야 하는 내용
#장르별 평균 가격과 최저·최고 가격
#0일 이내 시작하는 공연 중 가격 상위 5개
#공연 장소별 등록된 공연 수 순위

    
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
import re 
import time

    
service = Service(ChromeDriverManager().install())
options = Options()
#options.add_argument('--headless')
options.add_argument('--window-size=1920x1080')
options.add_argument('--lnag=ko_KR')
options.add_argument('--start-maximized')
options.add_argument('--user-agent=Mozilla/5.0 (Linux; Android 6.0; Nexus 5 Build/MRA58N) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Mobile Safari/537.36')
    
driver=webdriver.Chrome(service=service, options=options)
url = 'https://nol.interpark.com/'

driver.get(url)

driver.execute_script("window.scrollTo(0, document.body.scrollHeight)")
time.sleep(2)

tour_button = driver.find_element(By.LINK_TEXT, "티켓 홈 바로가기")
tour_button.click()
time.sleep(2)

concert_button = driver.find_element(By.LINK_TEXT, "콘서트")
concert_button.click()
time.sleep(2)


hot_button = driver.find_element(By.XPATH, "//button[@aria-label='요즘 HOT']")
hot_button.click()
time.sleep(5)

driver.execute_script("window.scrollTo(0, document.body.scrollHeight)")
time.sleep(2)

links = driver.find_elements(By.CSS_SELECTOR, "a.TicketItem_ticketItem__H51Vs")
detail_urls = []
for a in links:
    onclick_val = a.get_attribute("onclick")  # onclick="fnGoDetail('25007883')"
    if onclick_val:
        import re
        match = re.search(r"\d+", onclick_val)
        if match:
            code = match.group()
            detail_urls.append(f"https://tickets.interpark.com/contents/bridge/{code}")

print(f"총 {len(detail_urls)}개 링크 수집")
for url in detail_urls:
    print(url)

  
  
```
