```python
import instaloader
from collections import Counter
import pandas as pd

# -----------------------------
# 1) 인스타 크롤러 기본 설정
# -----------------------------
L = instaloader.Instaloader()

# 여성 감성층이 많은 핵심 태그들 (여기서부터 시작)
target_tags = [
    "감성툰", "감성일러스트", "일상툰",
    "위로툰", "그림일기", "감성그림"
]

max_posts_per_tag = 20  # 태그당 몇 개의 인기 게시물 가져올지

all_captions = []
all_hashtags = []

# -----------------------------
# 2) 태그별 게시물 수집
# -----------------------------
for tag in target_tags:
    print(f"📌 Collecting: #{tag}")
    posts = instaloader.Hashtag.from_name(L.context, tag).get_posts()

    for i, post in enumerate(posts):
        if i >= max_posts_per_tag:
            break
        
        caption = post.caption if post.caption else ""
        hashtags = post.caption_hashtags if post.caption else []

        all_captions.append(caption)
        all_hashtags.extend(hashtags)

# -----------------------------
# 3) 해시태그 반복도 분석
# -----------------------------
tag_counter = Counter(all_hashtags)
top_tags = tag_counter.most_common(50)  # 상위 50개만

df_tags = pd.DataFrame(top_tags, columns=["hashtag", "count"])

print("\n🔍 Top Hashtags:")
print(df_tags.head(20))

# -----------------------------
# 4) 감성형/템포형 자동 분류 로직
# -----------------------------
def classify_tag(tag):
    emo_keywords = ["감성", "몽글", "잔잔", "위로", "마음", "힐링", "따뜻", "공감", "파스텔", "연애"]
    tempo_keywords = ["병맛", "짤툰", "숏", "4컷", "반전", "웃긴"]

    if any(k in tag for k in emo_keywords):
        return "감성형"
    if any(k in tag for k in tempo_keywords):
        return "템포형"
    return "기타"

df_tags["segment"] = df_tags["hashtag"].apply(classify_tag)

print("\n🎨 감성형 / 템포형 자동 분류:")
print(df_tags.head(20))

# -----------------------------
# 5) CSV 저장
# -----------------------------
df_tags.to_csv("mumu_instagram_keyword_extract.csv", index=False, encoding='utf-8-sig')
print("\n📁 Saved → mumu_instagram_keyword_extract.csv")
```
