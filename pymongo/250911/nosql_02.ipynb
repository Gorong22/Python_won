```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017")

db = client.sample_mflix
movies = db.movies

print(movies)
```

```python
#'Star' 라는 단어가 포함된 영화 제목을 가져오세요 

import re
from pprint import pprint

regex = re.compile("Star", re.I)

for movie in movies.find({"title": regex}).limit(1) :
    pprint(movie)


#----- 선생님 코드 

star_movies = movies.find({"title":{"$regex":".*Star.*"}},{"title":1, "_id":0})

for movie in star_movies :
    print("영화제목:", movie["title"])
```

```python
#각 감독들이 제작한 영화의 수가 25개 이상인 감독들만 찾아서 출력해라 




#---- 선생님 코드 

director_Pipelines = [
    {"$unwind":"$directors"},
    {"$group": {"_id": "$directors", "count": {"$sum":1}}},
    {"$match":{"count":{"$gte": 25}}}
]

director_counts = list(movies.aggregate(director_Pipelines))

for director in director_counts:
    print(f"감독: {director['_id']}, 영화 수: {director['count']}편")
```

```python
# 평점을 기준으로 상위 5개의 영화를 찾으세요 (단 투표를 1000표 이상 받은 영화에 한함)


#선생님 코드 
top_rated_movies = list(
    movies.find({"imdb.votes": {"$gte": 1000}})
          .sort([("imdb.rating", -1)])
          .limit(5)
)

for movie in top_rated_movies:
    print(f"영화제목 : {movie['title']}, 평점 : {movie['imdb']['rating']}점")
```
