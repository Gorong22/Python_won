```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017")

db = client.sample_mflix
movies = db.movies

print(movies)
```

```python
# find() => aggregation Framework문법을 사용하면 pipeline 생성해서 복잡한 데이터처리를 순차적으로 세팅!

# 1. $match => 특정조건에 해당되는 문서를 찾아오고자 할 때
# 2. $limit => find().limit(2) 동일한 기능

for movie in movies.aggregate([{"$match": {"genres": "Action"}}, {"$limit": 1}]) :
    print(movie)
```

```python
# #group : SQL에서의 GROUP BY와 동일한 기능

for movie in movies.aggregate([{"$group": {"_id": "$director", "count": {"$sum": 1}}}]) :
    print(movie)
```
