```python
import requests
from bs4 import BeautifulSoup

from pymongo import MongoClient
client = MongoClient("mongodb://localhost:27017")
db = client["test"]
collection = db.daveshop

for page_num in range(10) :
    if page_num == 0 :
        res = requests.get("https://davelee-fun.github.io/")
    else : 
        res = requests.get(f"https://davelee-fun.github.io/page{str(page_num + 1)}")
    soup = BeautifulSoup(res.content, "html.parser")
    data = soup.select("div.card-body")
    for item in data :
        category = item.select_one("h2.card-title").get_text().replace("관련 상품 추천", "").strip()
        product = item.select_one("h4.card-text").get_text().replace("상품명:", "").strip()
        print(category, product)
        data = dict()
        data["category"] = category
        data["product"] = product
        collection.insert_one(data)
client.close()
```

```python
#모든 상품(product)를 조회하세요 선생님 답변 

product = collection.find({}, {"_id":0, "product" : 1})

for products in product :
    print(products["product"])
```

```python
#메트릭스 커버 카테고르의 모든 상품을 조회해라 

from pymongo import MongoClient
client = MongoClient("mongodb://localhost:27017")
db = client["test"]

mat = collection.find(
    {"category":"메트리스커버"},
    {"_id": 0, "product" :1}
    
)

for mats in mat :
    print(mats)
```

```python
from pymongo import MongoClient
client = MongoClient("mongodb://localhost:27017")
db = client["test"]
pipeline = [
    {"$match": {"category": "메트리스커버"}},
    {"$project": {"_id": 0, "product": 1}}
]

for product in collection.aggregate(pipeline):
    print(product)
```

```python
#각 카테고리별 상품수 조회 

pipeline = [
    {"$group":{"_id":"$category", "count":{"$sum":1}}},
     {"$project":{"_id":0, "categorty":"$_id","count":1}}

]

for product in collection.aggregate(pipeline):
    print(product)
```

```python
#상품명에 "핑크가 포함된 상품을 조회 해라"
pipeline = [
    {"$match":{"product":{"$regex":".*핑크.*", "$options":"i"}}},
    {"$project":{"_id":0, "product":1}}
    
]


for product in collection.aggregate(pipeline):
    print(product)
```
