```python
from pymongo import MongoClient

# 1. MongoClient 객체를 client 변수에 할당합니다.
client = MongoClient("mongodb://localhost:27017")

# 2. sample_mflix 데이터베이스를 db 변수에 할당합니다.
db = client["sample_mflix"]

# 3. 이제 db를 통해 컬렉션에 접근할 수 있습니다.
users = db.users
comments = db.comments
movies = db.movies

# 4. db.name을 출력하면 "sample_mflix"가 올바르게 나옵니다.
print("ok", db.name, "collections:",[c for c in db.list_collection_names() if c in {"users", "comments", "movies"}])
```

```python
loyal_users_Pipelines = [
    {
        "$lookup":{
            "from": "comments",
            "localField": "email",
            "foreignField": "email",
            "as": "c",
        }
    },
    {
        "$addFields":{
            "commentsCount": {"$size": "$c"},
            "avgTextLen": {
                "$avg": {
                    "$map":{
                        "input":"$c",
                        "as":"x",
                        "in": {"$strLenCP": {"$ifNull": ["$$x.text", ""]} }
                    }
                }
            },
             "lastCommentDate" : {"$max": "$c.date"}
        }
       
        
    },
    {"$match": {"commentsCount":{"$gte": 10}}},
    {"$sort": {"commentsCount":-1, "avgTextLen":-1, "lastCommentDate":-1}},
    {
        "$project": {
            "_id" : 0, 
            "name" : 1, 
            "email" : 1, 
            "commentsCount" : 1, 
            "avgTextLen" : 1, 
            "lastCommentDate" : 1 
     
        }
    }
]

loyal_users = list(users.aggregate(loyal_users_Pipelines))  #aggregation 프레임웍을 사용해야함 파이프라인 사용 하기 위함 
len(loyal_users), loyal_users[0:5]

#nosql -> 현업 실무 개발자들은 join을 싫어함 
#-> 서로 다른 데이터 테이블 OR 컬렉션을 하나로 연결해서 가져와 처리하려면 
#해당 데이터의 양이 방대해짐 
#방대해진 데이터를 처리하기위한 하드웨어에 많은 무리가 가고 그 무리가 간다는 건 
#기계에 대한 성능과 사양에 제약이 생기기 떄문에 좋아하지 않음 
```

```python
#유저,커멘트 연결 
#유저 -> 이메일 
#커멘트 -> text 길이 

movies_report_Pipeline = [
    {
        "$facet": {
            "latest5": [
                {"$sort": {"year":-1}},
                {"$limit":5},
                {"$project":{"_id":0, "title":1, "year":1}}
            ],
            "highRatedCount":[
                {"$match":{"imdb.rating":{"$gte":8}}},
                {"$count": "count"}
            ],
            "genresTop10": [
                {"$unwind": "$genres"},
                {"$group": {"_id":"$genres", "count":{"$sum":1}}},
                {"$sort": {"count":-1}},
                {"$limit": 10}
            ],
            "yearlyAvgRecent10":[
                {"$group":{"_id":"$year", "avgRating": {"$avg":"$imdb.rating"}}},
                {"$sort":{"_id":-1}},
                {"$limit":10},
                {"$sort":{"_id":-1}},
                {
                    "$project": {
                        "_id": 0,
                        "year" : "$_id",
                        "avgRating": 1
                        
                        
                    }
                }
            ]
            
        }
    }
]


movie_report = list(movies.aggregate(movies_report_Pipeline))
movie_report
```
