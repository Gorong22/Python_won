```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017")

db = client["test"]
collection = db.users
data = {"name": "Jhon", "age": 30}
result = collection.insert_one(data)
print("Inserted ID : ", result.inserted_id)

data = [
    {"name": "Alice", "age": 25},
    {"name": "Bob", "age": 35},
    {"name": "Charlie", "age": 40}
]

result = collection.insert_many(data)
print("Inserted IDs : ", result.inserted_ids)
```

```python
# find() => 모든 데이터를 찾아오게 된다면, list 자료구조!! || find_one()
```

```python
db = client["test"]
collection = db.users

document = collection.find_one({"name":"Bob"})
print(document)
```

```python
documents = collection.find()
for document in documents :
    print(document)
```

```python
# 데이터 문서 수정 : update_one() || update_many()
```

```python
collection = db.users

# collection.update_one("조건 = filter", "변경할 값 = update")
filter = {"name": "Jhon"}
update = {"$set": {"age": 31}}
result = collection.update_one(filter, update)

print("Modified Count : ", result.modified_count)
```

```python
filter = {"age": {"$gte": 40}}
update = {"$set": {"is_available": True}}
result = collection.update_many(filter, update)

print("Modified Count : ", result.modified_count)
```

```python
# delete_one() || delete_many()
```

```python
collection = db.users
filter = {"name": "Jhon"}
result = collection.delete_one(filter)

print("Deleted Count : ", result.deleted_count)
```

```python
collection = db.users
filter = {"age": {"$lt": 10}}
result = collection.delete_many(filter)

print("Deleted Count : ", result.deleted_count)
```

```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017")
db = client.sample_mflix
movies = db.movies

print(movies)
```

```python
# JS파일 포맷에서는 객체안에 있는 값을 찾아올 때, 대괄호 & 온점 표기법 가능
# 파이썬 포맷에서는 객체형태를 띄고있는 자료구조를 딕셔너리 => 대괄호 표기법으로 특정 키 내부의 값을 찾아온다

for movie in movies.find({"year": 1923}, {"_id": 0, "title": 1, "year": 1}) :
    print(movie, movie["year"])
```

```python
# MongoDB 안에 있는 특정 DB의 컬렉션 값을 어떤 조건과 함께 찾아오게 되면 기본적으로 리스트 형태를 띄고있음
# 해당 리스트 자료구조 안에있는 각각의 문서는 딕셔너리의 자료구조로 되어있다

for movie in movies.find(
    {"$or": [{"year": {"$lt": 1900}}, {"year": {"$gt": 2015}}]},
    {"_id": 0, "title": 1, "year": 1}
) :
    print(movie)
    
```

```python
# $all : 배열 내부에 있는 값이 조건값 모두를 충족하는 경우를 찾아오고 싶을때
for movie in movies.find(
    {"genres": {"$all": ["Action", "Sci-Fi", "Drama"]}},
    {"_id": 0, "title": 1, "year": 1, "genres": 1}
) : 
    print(movie)
```

```python
for movie in movies.find().sort("imdb.rating", -1).skip(3).limit(3) :
    print(movie)
```

```python
# 정규표현식

import re

regex = re.compile("Star", re.I)

for movie in movies.find({"title": regex}).limit(1) :
    print(movie)
```

```python
for movie in movies.find({"title": {"$regex": "Star", "$options": "i"}}).limit(1) :
    print(movie)
```

```python
distinct_genres = movies.distinct("genres")
print(distinct_genres)
```

```python
# $in : 배열 내 필드 값 중 하나만이라도 가지고 있다면 찾아올 수 있음!
for movie in movies.find({"genres": {"$in": ["Action", "Adventure"]}}).limit(3) :
    print(movie)
```

```python
# $exists : 존재하다 => 특정 필드가 문서에 존재하는지 여부에 따라서 문서를 선택해서 값을 찾아옴!
```

```python
for movie in movies.find({"writers": {"$exists": False}}).limit(3) :
    print(movie)
```

```python
count = movies.count_documents({"genres": {"$in": ["Action", "Adventure"]}})
print(count)
```
